```python
from apscheduler.schedulers.background import BackgroundScheduler
from datetime import datetime
import sqlite3

def connect_db():
    return sqlite3.connect('database.db')

def send_notifications():
    today = datetime.today().strftime('%Y-%m-%d')
    conn = connect_db()
    cursor = conn.cursor()
    cursor.execute("""
        SELECT u.user_id, f.name, c.description 
        FROM UserPreferences u
        JOIN Festivals f ON u.festival_id = f.festival_id
        JOIN Customs c ON u.custom_id = c.custom_id
        WHERE f.date = ?
    """, (today,))
    notifications = cursor.fetchall()
    conn.close()
    for notification in notifications:
        user_id, festival_name, custom_description = notification
        # Here, implement your notification sending logic, e.g., email, SMS, or push notifications
        print(f"Notify User {user_id}: {festival_name} - {custom_description}")

scheduler = BackgroundScheduler()
scheduler.add_job(send_notifications, 'interval', days=1)
scheduler.start()

if __name__ == '__main__':
    try:
        while True:
            pass
    except (KeyboardInterrupt, SystemExit):
        scheduler.shutdown()
```
